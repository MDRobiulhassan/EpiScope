````python
from pathlib import Path
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

ROOT = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)

spark = (
    SparkSession.builder
    .appName("EpiScope-Missing-Records")
    .master("local[*]")
    .config("spark.ui.enabled", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

try:
    bronze = spark.read.parquet(
        str(ROOT / "data/bronze/encounters_parquet")
    )
    clean = spark.read.parquet(
        str(ROOT / "data/silver/encounters")
    )
    quarantine = spark.read.parquet(
        str(ROOT / "data/silver/quarantine")
    )

    clean_ids = clean.select("encounter_id")
    quarantine_ids = quarantine.select("encounter_id")
    silver_ids = clean_ids.union(quarantine_ids).distinct()

    missing = bronze.join(
        silver_ids,
        on="encounter_id",
        how="left_anti",
    )

    print("Missing records:", missing.count())
    print("\nMissing records by age validity:")
    missing.groupBy(
        F.when(
            F.col("age").isNull()
            | (F.col("age") < 0)
            | (F.col("age") > 120),
            "Invalid age",
        ).otherwise("Valid age").alias("age_status")
    ).count().show()

    print("\nSample missing records:")
    missing.select(
        "encounter_id",
        "age",
        "temperature_c",
        "encounter_date",
    ).show(20, truncate=False)

finally:
    spark.stop()
````

````python
from pathlib import Path
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

ROOT = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)

spark = (
    SparkSession.builder
    .appName("EpiScope-Quarantine-Reconciliation")
    .master("local[*]")
    .config("spark.ui.enabled", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

try:
    bronze = spark.read.parquet(
        str(ROOT / "data/bronze/encounters_parquet")
    )

    quarantine = spark.read.parquet(
        str(ROOT / "data/silver/quarantine")
    )

    # Independently identify records that should fail age or temperature.
    invalid_age = (
        F.col("age").isNull()
        | (F.col("age") < 0)
        | (F.col("age") > 120)
    )

    invalid_temp = (
        F.col("temperature_c").isNull()
        | (F.col("temperature_c") < -20)
        | (F.col("temperature_c") > 60)
    )

    expected_bad = bronze.filter(invalid_age | invalid_temp)
    expected_ids = expected_bad.select("encounter_id").distinct()
    actual_ids = quarantine.select("encounter_id").distinct()

    missing_from_quarantine = expected_ids.join(
        actual_ids,
        on="encounter_id",
        how="left_anti",
    )

    unexpected_in_quarantine = actual_ids.join(
        expected_ids,
        on="encounter_id",
        how="left_anti",
    )

    print("Expected invalid records:", expected_ids.count())
    print("Actual quarantined records:", actual_ids.count())
    print("Expected invalid IDs absent from quarantine:",
          missing_from_quarantine.count())
    print("Quarantine IDs not failing age/temperature:",
          unexpected_in_quarantine.count())

    print("\nSample expected-invalid records absent from quarantine:")
    missing_from_quarantine.join(
        bronze, on="encounter_id", how="inner"
    ).select(
        "encounter_id", "age", "temperature_c"
    ).show(20, truncate=False)

finally:
    spark.stop()
````

````python
import json
from pathlib import Path

json_files = list(ROOT.rglob("*.json"))
print("JSON files found:", len(json_files))

errors = []
small_counts = []

def inspect(obj, path="$"):
    if isinstance(obj, dict):
        for key, value in obj.items():
            p = f"{path}.{key}"
            if isinstance(value, (int, float)) and not isinstance(value, bool):
                if any(term in key.lower() for term in ("case", "count", "incidence")):
                    if 0 < value < 5:
                        small_counts.append((p, value))
            inspect(value, p)
    elif isinstance(obj, list):
        for i, value in enumerate(obj):
            inspect(value, f"{path}[{i}]")

for file in json_files:
    try:
        with file.open(encoding="utf-8") as f:
            data = json.load(f)
        print(f"OK: {file.relative_to(ROOT)} ({file.stat().st_size:,} bytes)")
        if "public" in str(file).lower() or "dashboard" in str(file).lower():
            inspect(data, str(file.relative_to(ROOT)))
    except Exception as e:
        errors.append((str(file), str(e)))

print("\nInvalid JSON files:", len(errors))
for file, error in errors:
    print(file, error)

print("Potential unsuppressed counts 1–4:", len(small_counts))
for item in small_counts[:20]:
    print(item)

if not json_files:
    print("WARNING: No JSON exports found.")
````

````python

from pathlib import Path
from pyspark.sql import SparkSession, functions as F

ROOT = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)

spark = (
    SparkSession.builder
    .appName("EpiScope-ReadOnly-Diagnostics")
    .master("local[*]")
    .config("spark.ui.enabled", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

try:
    daily_path = str(ROOT / "data/gold/daily_cases")
    scores_path = str(ROOT / "data/gold/outbreak_scores")

    daily = spark.read.parquet(daily_path)
    scores = spark.read.parquet(scores_path)

    print("=== DAILY GOLD DATA ===")
    daily.agg(
        F.count("*").alias("daily_rows"),
        F.sum("case_count").alias("total_cases"),
        F.min("encounter_date").alias("first_date"),
        F.max("encounter_date").alias("last_date"),
        F.sum(
            F.when(F.col("ground_truth_outbreak") == True, 1).otherwise(0)
        ).alias("positive_label_rows"),
    ).show(truncate=False)

    print("=== SAVED DETECTION SCORES ===")
    scores.agg(
        F.count("*").alias("score_rows"),
        F.sum(
            F.when(F.col("is_outbreak_alert") == True, 1).otherwise(0)
        ).alias("alert_rows"),
        F.sum(
            F.when(F.col("baseline_days") >= 14, 1).otherwise(0)
        ).alias("eligible_rows"),
        F.max("case_count").alias("maximum_daily_cases"),
        F.max("baseline_days").alias("maximum_baseline_days"),
    ).show(truncate=False)

    print("=== LABEL / ALERT OVERLAP ===")
    truth = daily.select(
        "encounter_date", "district", "disease", "ground_truth_outbreak"
    ).dropDuplicates(["encounter_date", "district", "disease"])

    joined = scores.join(
        truth,
        on=["encounter_date", "district", "disease"],
        how="left",
    )

    joined.agg(
        F.count("*").alias("joined_rows"),
        F.sum(
            F.when(F.col("ground_truth_outbreak").isNull(), 1).otherwise(0)
        ).alias("missing_truth_rows"),
        F.sum(
            F.when(F.col("ground_truth_outbreak") == True, 1).otherwise(0)
        ).alias("positive_truth_rows"),
        F.sum(
            F.when(
                (F.col("is_outbreak_alert") == True)
                & (F.col("ground_truth_outbreak") == True), 1
            ).otherwise(0)
        ).alias("true_positives"),
        F.sum(
            F.when(
                (F.col("is_outbreak_alert") == True)
                & (F.col("ground_truth_outbreak") == False), 1
            ).otherwise(0)
        ).alias("false_positives"),
    ).show(truncate=False)

finally:
    spark.stop()
````

````python

from pathlib import Path
from pyspark.sql import SparkSession, functions as F

ROOT = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)

spark = (
    SparkSession.builder
    .appName("EpiScope-Detection-Verification")
    .master("local[*]")
    .config("spark.ui.enabled", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

try:
    daily = spark.read.parquet(
        str(ROOT / "data/gold/daily_cases")
    )

    scores = spark.read.parquet(
        str(ROOT / "data/gold/outbreak_scores")
    )

    truth = (
        daily.select(
            "encounter_date",
            "district",
            "disease",
            F.col("ground_truth_outbreak").cast("boolean")
        )
        .dropDuplicates(["encounter_date", "district", "disease"])
    )

    joined = (
        scores.alias("s")
        .join(
            truth.alias("t"),
            on=["encounter_date", "district", "disease"],
            how="left"
        )
        .withColumn(
            "actual_outbreak",
            F.coalesce(F.col("ground_truth_outbreak"), F.lit(False))
        )
        .filter(F.col("baseline_days") >= 14)
    )

    result = joined.agg(
        F.sum(F.when(
            F.col("is_outbreak_alert") & F.col("actual_outbreak"), 1
        ).otherwise(0)).alias("TP"),

        F.sum(F.when(
            F.col("is_outbreak_alert") & ~F.col("actual_outbreak"), 1
        ).otherwise(0)).alias("FP"),

        F.sum(F.when(
            ~F.col("is_outbreak_alert") & F.col("actual_outbreak"), 1
        ).otherwise(0)).alias("FN"),

        F.sum(F.when(
            ~F.col("is_outbreak_alert") & ~F.col("actual_outbreak"), 1
        ).otherwise(0)).alias("TN"),
    )

    print("=== VERIFIED COUNTS FROM SAVED DATA ===")
    result.show(truncate=False)

finally:
    spark.stop()
````

````python

from pathlib import Path
from pyspark.sql import SparkSession, Window
from pyspark.sql import functions as F
from datetime import timedelta

root = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)
input_path = root / "data/gold/daily_cases"

spark = (
    SparkSession.builder
    .appName("EpiScope-Forecast-Diagnostic")
    .master("local[*]")
    .config("spark.ui.enabled", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

try:
    daily = (
        spark.read.parquet(str(input_path))
        .select(
            F.to_date("encounter_date").alias("encounter_date"),
            "district",
            "disease",
            F.col("case_count").cast("double").alias("case_count"),
        )
        .groupBy("encounter_date", "district", "disease")
        .agg(F.sum("case_count").alias("case_count"))
    )

    print("=== CASE COUNT DISTRIBUTION ===")
    daily.select(
        F.count("*").alias("rows"),
        F.countDistinct("district").alias("districts"),
        F.countDistinct("disease").alias("diseases"),
        F.min("case_count").alias("min_cases"),
        F.expr("percentile_approx(case_count, 0.5)").alias("median_cases"),
        F.avg("case_count").alias("mean_cases"),
        F.max("case_count").alias("max_cases"),
    ).show(truncate=False)

    latest_date = daily.agg(F.max("encounter_date")).first()[0]
    test_start = latest_date - timedelta(days=27)
    history_start = test_start - timedelta(days=7)

    print(f"Latest date: {latest_date}")
    print(f"Test period: {test_start} to {latest_date}")
    print(f"Training period: {history_start} to {test_start - timedelta(days=1)}")

    print("\n=== TEST PERIOD CASE COUNTS ===")
    daily.filter(
        F.col("encounter_date") >= F.lit(test_start)
    ).select(
        F.count("*").alias("rows"),
        F.avg("case_count").alias("mean_cases"),
        F.max("case_count").alias("max_cases"),
    ).show(truncate=False)

    print("\n=== HISTORICAL TRAINING CASE COUNTS ===")
    daily.filter(
        (F.col("encounter_date") >= F.lit(history_start))
        & (F.col("encounter_date") < F.lit(test_start))
    ).select(
        F.count("*").alias("rows"),
        F.avg("case_count").alias("mean_cases"),
        F.max("case_count").alias("max_cases"),
    ).show(truncate=False)

finally:
    spark.stop()
````

````python
from pathlib import Path

report = r"""
\documentclass[conference]{IEEEtran}

\usepackage{cite}
\usepackage{amsmath}
\usepackage{booktabs}
\usepackage{array}
\usepackage{url}

\title{EpiScope: A Big Data Analytics Pipeline for Synthetic Disease Surveillance, Outbreak Detection, and Preparedness Support}

\author{
\IEEEauthorblockN{Robiul Hassan}
\IEEEauthorblockA{Department of Computer Science and Engineering\\
Premier University, Chattogram, Bangladesh\\
Email: \textit{[add university email]}}
}

\begin{document}
\maketitle

\begin{abstract}
Public-health analytics can require the processing of large encounter datasets, data-quality controls, trend analysis, and the preparation of information for operational planning. This paper presents EpiScope, a prototype big-data pipeline that demonstrates these tasks using synthetic disease-encounter data. The architecture uses Apache Spark for distributed-style data transformations, Bronze--Silver--Gold processing layers, Parquet analytical storage, baseline outbreak detection and forecasting, resource-allocation and preparedness-calendar modules, and a D3.js dashboard export path. In a documented 25-million-row run, 24,825,210 records were retained in the clean Silver output and 174,790 were quarantined, reconciling to the input count. Documented processing-stage timings total 27.40 minutes, excluding synthetic-data generation. The latest saved detection evaluation reports 190 true positives, 25,336 false positives, 169 false negatives, and 554,145 true negatives, corresponding to 52.92\% sensitivity, 4.37\% false-positive rate, and 0.74\% precision. The latest saved 28-day forecast evaluation reports a mean absolute error of 2.0182 and root mean squared error of 2.6986 across 8,960 district--disease--day observations. Results are limited to the synthetic dataset and current evaluation procedures. Historical metrics differ substantially and remain unreconciled. EpiScope is therefore presented as an educational and analytical prototype, not a validated public-health decision system.
\end{abstract}

\begin{IEEEkeywords}
big data analytics, Apache Spark, data quality, disease surveillance, outbreak detection, forecasting, privacy-aware visualization
\end{IEEEkeywords}

\section{Introduction}
Disease-surveillance workflows may involve large volumes of records from different sources and require reliable preparation before analytical results can be presented. A useful analytics prototype should demonstrate data ingestion, validation, aggregation, analytical processing, and the controlled publication of derived information.

EpiScope was developed as a big-data analytics project to demonstrate such a workflow on synthetic encounter records. It combines layered data processing with simple detection and forecasting baselines, resource-allocation and preparedness-calendar outputs, and a browser dashboard. The project emphasizes an end-to-end data path and records limitations in the evaluation and privacy checks.

The main contributions of this project are:
\begin{itemize}
    \item a Bronze--Silver--Gold processing workflow for synthetic encounter data;
    \item a documented large-input data-quality reconciliation and quarantine process;
    \item baseline outbreak-detection and fixed-horizon forecasting evaluations;
    \item experiments documenting pipeline timing, Gold-stage scaling, and Spark configuration comparisons; and
    \item a public-data preparation step that suppresses tested small case counts before dashboard export.
\end{itemize}

The paper does not claim clinical validity, real-world epidemiological accuracy, or formal privacy guarantees.

\section{System Architecture}
EpiScope separates data preparation, analytics, and dashboard serving. Synthetic encounter records are ingested into the Bronze layer. The Silver stage validates records and separates accepted rows from quarantined rows. The Gold stage creates aggregated analytical datasets. Detection, forecasting, resource allocation, and preparedness-calendar modules consume analytical outputs. Public-data preparation is then applied before JSON datasets are exported to the D3.js dashboard.

Apache Spark is used for data transformations, Parquet for processed analytical storage, and D3.js v7 for browser visualization. The project runs in a Google Colab environment with Spark in local mode for the recorded experiments. Consequently, measured runtimes depend on the runtime allocation, storage, partitioning, caching, and Spark configuration.

\section{Data and Processing Method}
\subsection{Synthetic Input}
The encounter records and outbreak labels used in the recorded experiments are synthetic. The project includes a generator for creating encounter records at different scales. Weather-related fields, if present in a given generated schema, must likewise be treated as synthetic or derived unless their provenance is independently established. Results should not be interpreted as evidence of performance on hospital records, government surveillance data, or real weather observations.

\subsection{Layered Processing}
The pipeline uses three logical data layers:
\begin{itemize}
    \item \textbf{Bronze:} ingestion of source encounter records;
    \item \textbf{Silver:} validation, cleaning, and quarantine of records that fail implemented checks;
    \item \textbf{Gold:} aggregated datasets for downstream analytics and serving.
\end{itemize}
The Silver stage aims to preserve accounting between accepted and quarantined rows. Reconciliation demonstrates count consistency but does not independently prove that every validation rule is correct.

\subsection{Analytics}
The outbreak detector is a prototype evaluated at the district--disease--day level for observations with at least 14 baseline days. Forecasting uses a simple seven-day-average baseline computed before the test period and repeated as a fixed prediction across a 28-day horizon. This is not a rolling forecast that updates predictions every day. Resource allocation and preparedness-calendar components are included as analytical demonstrations; their outputs are not validated operational guidance.

\subsection{Public Data and Privacy}
A public-data preparation step precedes dashboard JSON export. The documented policy suppresses exact case counts from one through four. A recorded check of the public case export found no exposed values in that range in the tested output. This is a limited check, not a complete privacy audit: inference may still be possible through geographic and temporal combinations, totals, linked views, or repeated queries.

\section{Experimental Setup}
The recorded experiments were run in Google Colab using Spark in local mode. The documented full-scale processing input contained 25,000,000 rows. Synthetic-data generation is excluded from the reported 27.40-minute processing-stage total. The report distinguishes current saved evaluation outputs from historical results that could not be independently reproduced from the audited artifacts.

Forecast evaluation used the test period from December 4 through December 31, 2025, with 8,960 district--disease--day observations. Detection metrics are taken from the latest saved evaluation output. Timing, optimization, scaling, and data-quality counts are historical documented measurements; raw timing logs were not available in the audited paths.

\section{Results}
\subsection{Data Quality}
Table~\ref{tab:quality} summarizes the documented reconciliation for the 25-million-row run.

\begin{table}[htbp]
\caption{Documented Silver-Stage Reconciliation}
\label{tab:quality}
\centering
\begin{tabular}{lr}
\toprule
Measure & Rows \\
\midrule
Input & 25,000,000 \\
Clean output & 24,825,210 \\
Quarantined output & 174,790 \\
Clean + quarantined & 25,000,000 \\
Quarantine rate & 0.69916\% \\
\bottomrule
\end{tabular}
\end{table}

The clean and quarantined counts sum to the input count. These values were transcribed from the project's documentation and were not independently recounted during report preparation.

\subsection{Pipeline Runtime}
Table~\ref{tab:runtime} presents documented stage runtimes. The total is a sum of recorded stages, not an independent timing measurement.

\begin{table}[htbp]
\caption{Documented Processing-Stage Runtime}
\label{tab:runtime}
\centering
\begin{tabular}{lr}
\toprule
Stage & Runtime (min) \\
\midrule
Bronze & 5.16 \\
Silver & 12.83 \\
Gold & 3.57 \\
Outbreak detection & 1.47 \\
Forecast generation & 0.94 \\
Resource allocation & 0.59 \\
Preparedness calendar & 0.57 \\
Public-data build & 0.50 \\
Dashboard export & 0.66 \\
Forecast evaluation & 0.86 \\
Detection evaluation & 0.75 \\
\midrule
Total measured stages & 27.40 \\
\bottomrule
\end{tabular}
\end{table}

The documented total is below the project's 45-minute target. Synthetic-data generation, recorded separately at approximately 7.17 minutes, is excluded. These measurements are environment-specific, and the raw timing logs were not found in the audited paths.

\subsection{Outbreak Detection}
Table~\ref{tab:detection} shows the latest saved detection evaluation.

\begin{table}[htbp]
\caption{Latest Saved Outbreak-Detection Evaluation}
\label{tab:detection}
\centering
\begin{tabular}{lr}
\toprule
Metric & Value \\
\midrule
True positives (TP) & 190 \\
False positives (FP) & 25,336 \\
False negatives (FN) & 169 \\
True negatives (TN) & 554,145 \\
Sensitivity & 52.92\% \\
False-positive rate & 4.37\% \\
Precision & 0.74\% \\
\bottomrule
\end{tabular}
\end{table}

Sensitivity indicates that the detector identifies approximately half of the positive evaluation cases. However, precision of 0.74\% means that false-positive alerts greatly outnumber true-positive alerts. This substantially limits the usefulness of the current detector without review and further tuning. It must not be used for unreviewed operational decisions.

\subsection{Forecasting}
The latest saved forecast evaluation contains 8,960 observations for December 4--31, 2025. The seven-day-average fixed-horizon baseline achieved MAE \(=2.0182\) cases and RMSE \(=2.6986\) cases. MAE expresses the average absolute difference between predicted and actual counts; RMSE weights larger errors more heavily. These values describe only the current synthetic evaluation and depend on the evaluation target and calendar-completion procedure.

Earlier project documentation reported MAE approximately 5.9894 and RMSE approximately 11.6049. The original run and configuration for those values were not independently verified, and the discrepancy remains unresolved. The latest saved output is reported here as the current result, while the earlier figures are retained as unverified historical results.

\subsection{Optimization and Scaling}
Documented Gold-stage experiments report 4.91 minutes for AQE disabled with 200 shuffle partitions, compared with 3.45 minutes for AQE enabled with 32 partitions. The latter is approximately 29.7\% lower than the 4.91-minute baseline. Because these are historical single-run measurements without raw logs in the audited paths, they should be interpreted as indicative observations rather than a reliable general speedup.

Gold-stage runtimes were documented as 1.55 minutes for 6.25 million rows, 2.03 minutes for 12.5 million rows, and 3.45 minutes for 24,825,210 clean rows. The smaller datasets were generated using prefix or row-limit samples and may not preserve the full dataset distribution. These measurements describe the Gold stage only, not end-to-end pipeline scaling.

\section{Discussion}
The project demonstrates an integrated workflow that connects large-input processing, data-quality reconciliation, baseline analytics, and dashboard-oriented export. The documented Silver-stage reconciliation provides evidence of row accounting for the recorded run. The runtime records suggest that the pipeline stages fit within the stated processing target in the measured environment, although the absence of raw timing logs limits independent verification.

The analytical results also show why evaluation metrics must be interpreted together. A sensitivity of 52.92\% may appear useful in isolation, but precision of 0.74\% indicates a high false-alert burden. Similarly, the forecast errors cannot establish real-world predictive utility because the evaluation data are synthetic and the baseline is intentionally simple. The unresolved difference between current and historical forecast metrics further motivates stronger experiment provenance and automated result capture.

\section{Limitations and Future Work}
Several limitations remain:
\begin{itemize}
    \item \textbf{Synthetic data:} results do not establish performance on real surveillance data.
    \item \textbf{Detection quality:} very low precision and a substantial false-positive count require investigation of label construction, thresholds, class imbalance, and evaluation logic.
    \item \textbf{Forecast baseline:} the fixed seven-day average does not adapt through the 28-day horizon and should be compared against additional baselines using a reproducible evaluation protocol.
    \item \textbf{Historical discrepancies:} older forecast and detection figures differ from current saved outputs and have not been reconciled.
    \item \textbf{Benchmark evidence:} historical runtime and optimization records lack raw logs in the audited paths, and the reported values may be affected by caching and environment variability.
    \item \textbf{Privacy:} suppressing small counts is not a formal privacy guarantee. All public exports and cross-view inference risks require further testing.
    \item \textbf{Quality-rule evidence:} the distinct implemented Silver validation rules and their test evidence should be inventoried from source code before claiming full compliance with a required rule count.
\end{itemize}

Future work should include reproducible experiment logging, repeated benchmark runs, a source-derived data-quality rule inventory, expanded public-output privacy tests, alternative forecasting baselines, and detector threshold analysis. Any application to real public-health decisions would require representative real data, domain-expert review, and appropriate governance.

\section{Conclusion}
EpiScope demonstrates a layered big-data pipeline for synthetic disease-surveillance analytics, including data cleaning and quarantine, aggregation, baseline detection and forecasting, and dashboard data preparation. The documented large-input run reconciles 25 million input rows and reports 27.40 minutes across measured processing and evaluation stages, excluding data generation. The latest saved evaluations report 52.92\% detection sensitivity but only 0.74\% precision, and forecast MAE/RMSE of 2.0182/2.6986 on a synthetic 28-day test period. These results are preliminary and subject to unresolved historical discrepancies and limitations in independent reproducibility. The system should be regarded as an educational prototype rather than a validated public-health decision-support tool.

\begin{thebibliography}{00}
\bibitem{spark}
Apache Software Foundation, ``Apache Spark Documentation.'' [Online]. Available: \url{https://spark.apache.org/docs/latest/}

\bibitem{parquet}
Apache Software Foundation, ``Apache Parquet Documentation.'' [Online]. Available: \url{https://parquet.apache.org/docs/}

\bibitem{d3}
M. Bostock, V. Ogievetsky, and J. Heer, ``D$^3$: Data-Driven Documents,'' \emph{IEEE Transactions on Visualization and Computer Graphics}, vol. 17, no. 12, pp. 2301--2309, Dec. 2011, doi: 10.1109/TVCG.2011.185.

\bibitem{ieee}
IEEE, ``IEEE Reference Guide.'' [Online]. Available: \url{https://journals.ieeeauthorcenter.ieee.org/your-role-in-article-production/ieee-editorial-style-manual/}
\end{thebibliography}

\end{document}
""".strip() + "\n"

from pathlib import Path

out = Path("/mnt/data/episcope_ieee_report.tex")
out.parent.mkdir(parents=True, exist_ok=True)
out.write_text(report, encoding="utf-8")

print(f"Created: {out}")
print(f"Characters: {len(report):,}")
````

````python
import json
from pathlib import Path

project = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)
public_dir = project / "dashboard" / "data"

files_to_check = [
    "cases.json",
    "alerts.json",
    "forecasts.json",
    "resources.json",
    "calendar.json",
]

# Only fields whose names indicate a count are tested.
# Extend this list if your schema uses other count field names.
def is_count_field(key):
    key = key.lower()
    return (
        "count" in key
        or key in {"cases", "case", "total", "total_cases", "case_count"}
    )

for filename in files_to_check:
    path = public_dir / filename

    print(f"\n{'=' * 60}\n{filename}")

    if not path.exists():
        print("MISSING FILE")
        continue

    try:
        data = json.loads(path.read_text(encoding="utf-8"))
    except (json.JSONDecodeError, UnicodeDecodeError) as exc:
        print(f"INVALID JSON: {exc}")
        continue

    findings = []

    def walk(obj, location="$"):
        if isinstance(obj, dict):
            for key, value in obj.items():
                current = f"{location}.{key}"

                if (
                    is_count_field(key)
                    and isinstance(value, (int, float))
                    and not isinstance(value, bool)
                    and value in (1, 2, 3, 4, 5)
                ):
                    findings.append((current, value))

                walk(value, current)

        elif isinstance(obj, list):
            for i, value in enumerate(obj):
                walk(value, f"{location}[{i}]")

    walk(data)

    small = [(p, v) for p, v in findings if v in (1, 2, 3, 4)]
    fives = [(p, v) for p, v in findings if v == 5]

    print(f"Small count values (1–4): {len(small)}")
    for item in small[:20]:
        print("  REVIEW:", item)

    print(f"Count values equal to 5: {len(fives)}")
    for item in fives[:10]:
        print("  Boundary:", item)

    if not small:
        print("No explicit count fields with values 1–4 found.")
````

````python
SUPPRESSION_THRESHOLD = 5

test_counts = range(7)

print(f"{'Input':>8} {'Suppressed':>12} {'Published':>12} {'Expected':>12}")
print("-" * 48)

passed = True

for count in test_counts:
    suppressed = (
        count > 0 and count < SUPPRESSION_THRESHOLD
    )
    published = None if suppressed else count

    expected_suppressed = count in (1, 2, 3, 4)
    expected_published = None if expected_suppressed else count

    ok = (
        suppressed == expected_suppressed
        and published == expected_published
    )
    passed &= ok

    print(
        f"{count:>8} {str(suppressed):>12} "
        f"{str(published):>12} {'PASS' if ok else 'FAIL':>12}"
    )

print("\nRESULT:", "ALL TESTS PASSED" if passed else "TESTS FAILED")
````
