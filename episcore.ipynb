```python
from pathlib import Path
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

ROOT = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)

spark = (
    SparkSession.builder
    .appName("EpiScope-Validation")
    .getOrCreate()
)

paths = {
    "Bronze": ROOT / "data/bronze/encounters_parquet",
    "Silver clean": ROOT / "data/silver/encounters",
    "Silver quarantine": ROOT / "data/silver/quarantine",
    "Gold daily cases": ROOT / "data/gold/daily_cases",
}

for name, path in paths.items():
    print(f"\n--- {name} ---")

    if not path.exists() or not list(path.rglob("*.parquet")):
        print("Missing or empty Parquet output")
        continue

    df = spark.read.parquet(str(path))
    print("Rows:", f"{df.count():,}")
    print("Columns:", df.columns)

    if "encounter_date" in df.columns:
        df.select(
            F.min("encounter_date").alias("start_date"),
            F.max("encounter_date").alias("end_date")
        ).show(truncate=False)

spark.stop()
```

```python
from pathlib import Path
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

ROOT = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)

spark = (
    SparkSession.builder
    .appName("EpiScope-Missing-Records")
    .master("local[*]")
    .config("spark.ui.enabled", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

try:
    bronze = spark.read.parquet(
        str(ROOT / "data/bronze/encounters_parquet")
    )
    clean = spark.read.parquet(
        str(ROOT / "data/silver/encounters")
    )
    quarantine = spark.read.parquet(
        str(ROOT / "data/silver/quarantine")
    )

    clean_ids = clean.select("encounter_id")
    quarantine_ids = quarantine.select("encounter_id")
    silver_ids = clean_ids.union(quarantine_ids).distinct()

    missing = bronze.join(
        silver_ids,
        on="encounter_id",
        how="left_anti",
    )

    print("Missing records:", missing.count())
    print("\nMissing records by age validity:")
    missing.groupBy(
        F.when(
            F.col("age").isNull()
            | (F.col("age") < 0)
            | (F.col("age") > 120),
            "Invalid age",
        ).otherwise("Valid age").alias("age_status")
    ).count().show()

    print("\nSample missing records:")
    missing.select(
        "encounter_id",
        "age",
        "temperature_c",
        "encounter_date",
    ).show(20, truncate=False)

finally:
    spark.stop()
```

```python
from pathlib import Path
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

ROOT = Path(
    "/content/drive/MyDrive/University/8th Semester/BDA/episcope"
)

spark = (
    SparkSession.builder
    .appName("EpiScope-Quarantine-Reconciliation")
    .master("local[*]")
    .config("spark.ui.enabled", "false")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("ERROR")

try:
    bronze = spark.read.parquet(
        str(ROOT / "data/bronze/encounters_parquet")
    )

    quarantine = spark.read.parquet(
        str(ROOT / "data/silver/quarantine")
    )

    # Independently identify records that should fail age or temperature.
    invalid_age = (
        F.col("age").isNull()
        | (F.col("age") < 0)
        | (F.col("age") > 120)
    )

    invalid_temp = (
        F.col("temperature_c").isNull()
        | (F.col("temperature_c") < -20)
        | (F.col("temperature_c") > 60)
    )

    expected_bad = bronze.filter(invalid_age | invalid_temp)
    expected_ids = expected_bad.select("encounter_id").distinct()
    actual_ids = quarantine.select("encounter_id").distinct()

    missing_from_quarantine = expected_ids.join(
        actual_ids,
        on="encounter_id",
        how="left_anti",
    )

    unexpected_in_quarantine = actual_ids.join(
        expected_ids,
        on="encounter_id",
        how="left_anti",
    )

    print("Expected invalid records:", expected_ids.count())
    print("Actual quarantined records:", actual_ids.count())
    print("Expected invalid IDs absent from quarantine:",
          missing_from_quarantine.count())
    print("Quarantine IDs not failing age/temperature:",
          unexpected_in_quarantine.count())

    print("\nSample expected-invalid records absent from quarantine:")
    missing_from_quarantine.join(
        bronze, on="encounter_id", how="inner"
    ).select(
        "encounter_id", "age", "temperature_c"
    ).show(20, truncate=False)

finally:
    spark.stop()
```

```python
import json
from pathlib import Path

json_files = list(ROOT.rglob("*.json"))
print("JSON files found:", len(json_files))

errors = []
small_counts = []

def inspect(obj, path="$"):
    if isinstance(obj, dict):
        for key, value in obj.items():
            p = f"{path}.{key}"
            if isinstance(value, (int, float)) and not isinstance(value, bool):
                if any(term in key.lower() for term in ("case", "count", "incidence")):
                    if 0 < value < 5:
                        small_counts.append((p, value))
            inspect(value, p)
    elif isinstance(obj, list):
        for i, value in enumerate(obj):
            inspect(value, f"{path}[{i}]")

for file in json_files:
    try:
        with file.open(encoding="utf-8") as f:
            data = json.load(f)
        print(f"OK: {file.relative_to(ROOT)} ({file.stat().st_size:,} bytes)")
        if "public" in str(file).lower() or "dashboard" in str(file).lower():
            inspect(data, str(file.relative_to(ROOT)))
    except Exception as e:
        errors.append((str(file), str(e)))

print("\nInvalid JSON files:", len(errors))
for file, error in errors:
    print(file, error)

print("Potential unsuppressed counts 1–4:", len(small_counts))
for item in small_counts[:20]:
    print(item)

if not json_files:
    print("WARNING: No JSON exports found.")
```

```python
import json
from collections import Counter

path = ROOT / "dashboard" / "data" / "cases.json"
data = json.loads(path.read_text(encoding="utf-8"))

counts = Counter()
violations = []
suppressed_rows = 0

for row in data:
    count = row.get("published_case_count")
    suppressed = row.get("suppressed")

    if suppressed:
        suppressed_rows += 1

    if isinstance(count, (int, float)):
        if 0 < count < 5:
            counts[count] += 1
            violations.append(row)

print("Total public records:", len(data))
print("Suppressed records:", suppressed_rows)
print("Published counts 1–4:", dict(counts))
print("Privacy suppression check:", "PASS" if not violations else "FAIL")

if violations:
    print("\nFirst 10 potential violations:")
    for row in violations[:10]:
        print(row)
```

```python

from pathlib import Path
import re

project = ROOT

print("PROJECT:", project)
print("\n=== Relevant project files ===")

for folder in ["pipeline", "data", "config", "scripts", "analytics"]:
    base = project / folder
    if base.exists():
        for path in sorted(base.rglob("*")):
            if path.is_file() and path.suffix in {
                ".py", ".json", ".yaml", ".yml", ".toml", ".sh"
            }:
                print(path.relative_to(project))

print("\n=== Dataset size / scaling references ===")

patterns = re.compile(
    r"scale.factor|scalefactor|scale_factor|25_?000_?000|"
    r"3000000|3_000_000|encounter.count|num.records|"
    r"num_records|n_records|record_count",
    re.IGNORECASE
)

for path in project.rglob("*"):
    if not path.is_file() or path.suffix not in {".py", ".json", ".yaml", ".yml", ".toml", ".sh"}:
        continue
    try:
        if path.stat().st_size > 2_000_000:
            continue
        lines = path.read_text(encoding="utf-8").splitlines()
        matches = [
            (i, line.strip())
            for i, line in enumerate(lines, 1)
            if patterns.search(line)
        ]
        if matches:
            print(f"\n--- {path.relative_to(project)} ---")
            for i, line in matches[:15]:
                print(f"{i}: {line[:200]}")
    except (UnicodeDecodeError, OSError):
        pass
```

```python

from pathlib import Path
import shutil
import subprocess
import sys
import time

source_script = ROOT / "generator" / "generate_encounters.py"
local_script = Path("/content/generate_encounters_25m.py")
local_output = Path("/content/episcope_25m")

# Keep the existing project dataset untouched.
if local_output.exists():
    shutil.rmtree(local_output)

script_text = source_script.read_text(encoding="utf-8")
old_output = 'OUTPUT_DIR = ROOT / "data" / "bronze" / "encounters"'
new_output = 'OUTPUT_DIR = Path("/content/episcope_25m")'

if old_output not in script_text:
    raise RuntimeError("Generator output definition differs from expected. No generation started.")

local_script.write_text(
    script_text.replace(old_output, new_output, 1),
    encoding="utf-8"
)

start = time.perf_counter()

result = subprocess.run(
    [
        sys.executable, "-u", str(local_script),
        "--rows", "25000000",
        "--scale-factor", "1",
        "--chunk-size", "100000",
        "--seed", "42"
    ],
    text=True
)

print(f"\nGeneration runtime: {(time.perf_counter() - start) / 60:.2f} minutes")
print("Exit code:", result.returncode)

generated_csv = local_output / "encounters.csv"

if result.returncode != 0 or not generated_csv.exists():
    raise RuntimeError("Generation failed. The existing project dataset remains untouched.")

with generated_csv.open("rb") as f:
    row_count = sum(1 for _ in f) - 1  # Exclude CSV header

print(f"Verified data rows: {row_count:,}")
print(f"Generated CSV size: {generated_csv.stat().st_size / (1024**3):.2f} GiB")

if row_count != 25_000_000:
    raise RuntimeError("Row count does not match the 25-million-record requirement.")

print("GENERATION VERIFIED: PASS")
```
